```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_excel('Lat_Long_Data.xlsx')
data
```

```python
# Function to fix the direction anomaly and create a new column
def fix_direction(row):
    new_lat = row['Latitude']
    new_long = row['Longitude']
    
    # Check if latitude and longitude need to be flipped
    if row['Latitude'][-1] not in ('N', 'S') and row['Longitude'][-1] not in ('E', 'W'):
        new_lat = row['Longitude']
        new_long = row['Latitude']
        fix_action = 'Direction exchanged to fix the anomaly'
    else:
        fix_action = ''
    
    return pd.Series([new_lat, new_long, fix_action], index=['Latitude', 'Longitude', 'Fix Action'])

# Apply the function to each row
data[['Latitude', 'Longitude', 'Fix Action']] = data.apply(fix_direction, axis=1)
```
