```python
import numpy as np
import pandas as pd
```

# Data Type	Quantity Range
int8	-128 to 127

int16	-32,768 to 32,767

int32	-2,147,483,648 to 2,147,483,647

int64	-9,223,372,036,854,775,808 to 9,223,372,036,854,775,807

float16	Approximately ±6.103515625 × 10^−5 to ±65504

float32	Approximately ±1.40129846432481707 × 10^−45 to ±3.40282346638528860 × 10^38

float64	Approximately ±2.2250738585072014 × 10^−308 to ±1.7976931348623157 × 10^308

```python
# Convert col1 from object to integer
df['Quantity'] = df['Quantity'].astype(np.int16)

# Convert col1 from integer to float
df['Discount'] = df['Discount'].astype(np.float32)
```
