```python
import numpy as np
import pandas as pd
```

```python
# Function to split quantity and unit from weight string
def split_weight(weight_str):
    quantity, unit = weight_str.split(' ')
    return float(quantity), unit
```

```python
df_data = pd.read_excel('Conversion_data.xlsx')
df_data.head(3)
```

```python
df_unit_conversion = pd.read_excel('Unit_Conversion.xlsx')
df_unit_conversion
```

```python
# Apply split_weight function to df_data table
df_data[['Quantity', 'Unit']] = df_data['Weight'].apply(lambda x: pd.Series(split_weight(x)))
df_data.head(3)
```

```python
# Merge df_data with unit_conversion table
merged_table = pd.merge(df_data, df_unit_conversion, left_on='Unit', right_on='To', how='left')
merged_table
```

```python
# Calculate weights in kilograms
merged_table['Weight_in_kg'] = merged_table['Quantity'] / merged_table['Conversion Factor']
```
