```python
import numpy as np
import pandas as pd
```

```python
df_duplicates = pd.read_excel('Downloads/Duplicates.xlsx')
df_duplicates.head(3)
```

```python
# Add a new column to flag duplicates
df_duplicates['IsDuplicate'] = df_duplicates.duplicated(subset=['Name', 'City'], keep=False)
```

```python
df_duplicates.head(3)
```

```python
df_duplicates = df_duplicates[df_duplicates['IsDuplicate'] == True]
df_duplicates
```

```python
df_duplicates = df_duplicates.sort_values(by=['Name', 'City'])
df_duplicates
```
