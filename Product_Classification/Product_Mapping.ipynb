```python
# Load data from Excel files
file1 = pd.read_excel("S_Data.xlsx")
file2 = pd.read_excel("T_Data.xlsx")
```

```python
# Task 1: Inner join the two dataframes
joined_table = pd.merge(file2, file1, how="left", left_on="T_Commodity Title", right_on="S_Commodity Title")
```

```python
# Task 2: Create observations column
def classify(row):
    if pd.isna(row['S_Commodity Title']):
        return "Commodity Title not in Standard Table"
    elif pd.isna(row['T_Commodity Code']):
        return "Unclassified"
    elif row['S_Commodity Code'] != row['T_Commodity Code']:
        return "Misclassified"
    else:
        return "Rightly Classified"

joined_table['Observations'] = joined_table.apply(classify, axis=1)
```

```python
joined_table
```

```python
# Task 3: Create suggestions column
def suggest(row):
    if pd.isna(row['T_Commodity Code']):
        return row['S_Commodity Code']
    elif row['S_Commodity Code'] != row['T_Commodity Code']:
        return row['S_Commodity Code']
    else:
        return row['T_Commodity Code']

joined_table['Suggestions'] = joined_table.apply(suggest, axis=1)
```

```python
joined_table
```

```python
# Filter out rows that are rightly classified
filtered_table = joined_table[joined_table['Observations'] != 'Rightly Classified']

# Display the filtered table
filtered_table
```
